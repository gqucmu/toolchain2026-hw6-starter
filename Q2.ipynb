`data/jena_normalized.npz` holds the Jena Climate data, already split by time (2009-2014 train, 2015 validation,

of shape (rows, 14), one row every 10 minutes, with the 14 features in the order of `FEATURES`.

```python
import time

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader

torch.manual_seed(0)

jena = np.load("data/jena_normalized.npz")
train_data, validate_data, test_data = jena["train"], jena["validate"], jena["test"]
FEATURES = list(jena["features"])
TEMP = FEATURES.index("T (degC)")                    # column index of the temperature
T_mean, T_std = jena["mean"][TEMP], jena["std"][TEMP]  # temperature in degC = normalized value * T_std + T_mean

print(FEATURES)
for name, d in [("train", train_data), ("validate", validate_data), ("test", test_data)]:
    print(f"{name:>8}: {d.shape}, {d.dtype}")
print(f"temperature: mean {T_mean:.2f} degC, std {T_std:.2f} degC (training split)")
```
